```python
import torch
import cv2
import numpy as np
import os
from PIL import Image

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)

# Load MiDaS model
midas = torch.hub.load("intel-isl/MiDaS", "DPT_Hybrid")
midas.to(DEVICE).eval()

# Load MiDaS transforms
midas_transforms = torch.hub.load("intel-isl/MiDaS", "transforms")
transform = midas_transforms.dpt_transform

# Paths
IMG_DIR = "/content/drive/MyDrive/controlnet_project/images"
OUT_DIR = "/content/drive/MyDrive/controlnet_project/depth_maps"
os.makedirs(OUT_DIR, exist_ok=True)

# Load one image
img_files = os.listdir(IMG_DIR)
img_path = os.path.join(IMG_DIR, img_files[0])

# PIL → NumPy
img = Image.open(img_path).convert("RGB")
img = np.array(img)

# ✅ TRANSFORM (already returns correct tensor)
input_batch = transform(img).to(DEVICE)

# Inference
with torch.no_grad():
    prediction = midas(input_batch)

# Post-process
depth = prediction.squeeze().cpu().numpy()
depth = (depth - depth.min()) / (depth.max() - depth.min())

out_path = os.path.join(OUT_DIR, img_files[0])
cv2.imwrite(out_path, (depth * 255).astype(np.uint8))

print("Saved depth map:", out_path)
```

```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

IMG_DIR = "/content/drive/MyDrive/controlnet_project/images"
OUT_DIR = "/content/drive/MyDrive/controlnet_project/gaze_maps"
os.makedirs(OUT_DIR, exist_ok=True)

# Create saliency detector
saliency = cv2.saliency.StaticSaliencySpectralResidual_create()

# Optional: fixed size for ControlNet (recommended)
TARGET_SIZE = (512, 512)

img_files = sorted(os.listdir(IMG_DIR))

for fname in img_files:
    img_path = os.path.join(IMG_DIR, fname)

    # Read image safely
    img = cv2.imread(img_path)
    if img is None:
        print(f"Skipping non-image file: {fname}")
        continue

    # Resize for consistency
    img_resized = cv2.resize(img, TARGET_SIZE)

    # Compute saliency
    success, sal_map = saliency.computeSaliency(img_resized)
    if not success:
        print(f"Saliency failed for: {fname}")
        continue

    # Convert to uint8
    sal_map = (sal_map * 255).astype(np.uint8)

    # Smooth + normalize (IMPORTANT)
    sal_map = cv2.GaussianBlur(sal_map, (31, 31), 0)
    sal_map = cv2.normalize(sal_map, None, 0, 255, cv2.NORM_MINMAX)

    # Save
    out_path = os.path.join(OUT_DIR, fname)
    cv2.imwrite(out_path, sal_map)

    print("Saved gaze map:", fname)
```

```python
import os
import random
import requests
from tqdm import tqdm

OI_URLS_FILE = "https://storage.googleapis.com/openimages/2018_04/validation/validation-images-with-rotation.csv"

NUM_IMAGES = 300   # 👈 change as needed

os.makedirs(OI_DIR, exist_ok=True)

lines = requests.get(OI_URLS_FILE).text.strip().split("\n")[1:]
random.shuffle(lines)
lines = lines[:NUM_IMAGES]

for line in tqdm(lines):
    parts = line.split(",")
    img_id = parts[0]
    url = parts[2]

    out_path = os.path.join(OI_DIR, img_id + ".jpg")

    if os.path.exists(out_path):
        continue

    try:
        img_data = requests.get(url, timeout=10).content
        with open(out_path, "wb") as f:
            f.write(img_data)
    except:
        continue

print("✔ OpenImages subset downloaded")
```

```python
import torch
import cv2
import os
import numpy as np
from PIL import Image

# -------------------------
# PATHS
# -------------------------
BASE_DIR = "/content/drive/MyDrive/controlnet_project"
RGB_DIR = os.path.join(BASE_DIR, "rgb")
OUT_DIR = os.path.join(BASE_DIR, "depth_maps")

# -------------------------
# DEVICE
# -------------------------
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)


# -------------------------
# LOAD MIDAS
# -------------------------
midas = torch.hub.load(
    "intel-isl/MiDaS",
    "DPT_Hybrid",
    trust_repo=True
)
midas.to(DEVICE).eval()

midas_transforms = torch.hub.load(
    "intel-isl/MiDaS",
    "transforms"
)
transform = midas_transforms.dpt_transform

# -------------------------
# PROCESS FUNCTION
# -------------------------
def process_subset(subset):
    in_dir = os.path.join(RGB_DIR, subset)
    out_dir = os.path.join(OUT_DIR, subset)
    os.makedirs(out_dir, exist_ok=True)

    for fname in os.listdir(in_dir):
        img_path = os.path.join(in_dir, fname)

        # --- Load image ---
        img = Image.open(img_path).convert("RGB")
        img_np = np.array(img)   # 🔑 PIL → NumPy

        # --- Apply MiDaS transform ---
        input_batch = transform(img_np)
        input_batch = input_batch.to(DEVICE)

        # --- Inference ---
        with torch.no_grad():
            prediction = midas(input_batch)

        # --- Normalize depth ---
        depth = prediction.squeeze().cpu().numpy()
        depth = (depth - depth.min()) / (depth.max() - depth.min())

        # --- Save ---
        cv2.imwrite(
            os.path.join(out_dir, fname),
            (depth * 255).astype(np.uint8)
        )

        print(f"[DEPTH] {subset}/{fname}")

# -------------------------
# RUN
# -------------------------
process_subset("coco")
# process_subset("openimages")
```

```python
import os, json, shutil

BASE_DIR = "/content/drive/MyDrive/controlnet_project"

RGB_DIR = os.path.join(BASE_DIR, "rgb/coco")
DEPTH_DIR = os.path.join(BASE_DIR, "depth_maps/coco")

OUT_DIR = os.path.join(BASE_DIR, "train_depth")
IMG_OUT = os.path.join(OUT_DIR, "images")
COND_OUT = os.path.join(OUT_DIR, "conditioning")

os.makedirs(IMG_OUT, exist_ok=True)
os.makedirs(COND_OUT, exist_ok=True)

captions = {}

for fname in os.listdir(RGB_DIR):
    if not fname.lower().endswith(".jpg"):
        continue

    rgb_path = os.path.join(RGB_DIR, fname)
    depth_path = os.path.join(DEPTH_DIR, fname)

    # skip if depth image is missing
    if not os.path.exists(depth_path):
        print(f"⚠️ Missing depth map for {fname}, skipping")
        continue

    shutil.copy(rgb_path, os.path.join(IMG_OUT, fname))
    shutil.copy(depth_path, os.path.join(COND_OUT, fname))

    captions[fname] = "a realistic scene"

with open(os.path.join(OUT_DIR, "captions.json"), "w") as f:
    json.dump(captions, f, indent=2)

print("✔ Depth training dataset ready")
```

```python
import os, json, shutil

BASE_DIR = "/content/drive/MyDrive/controlnet_project"

RGB_DIR = os.path.join(BASE_DIR, "rgb/coco")
GAZE_DIR = os.path.join(BASE_DIR, "gaze_maps/coco")  # 🔹 FIXED PATH

OUT_DIR = os.path.join(BASE_DIR, "train_gaze")
IMG_OUT = os.path.join(OUT_DIR, "images")
COND_OUT = os.path.join(OUT_DIR, "conditioning")

assert os.path.exists(RGB_DIR), f"Missing RGB_DIR: {RGB_DIR}"
assert os.path.exists(GAZE_DIR), f"Missing GAZE_DIR: {GAZE_DIR}"

os.makedirs(IMG_OUT, exist_ok=True)
os.makedirs(COND_OUT, exist_ok=True)

# Build gaze lookup
gaze_lookup = {
    os.path.splitext(f)[0]: f
    for f in os.listdir(GAZE_DIR)
}

captions = {}
skipped = 0

for fname in os.listdir(RGB_DIR):
    if not fname.lower().endswith(".jpg"):
        continue

    base = os.path.splitext(fname)[0]

    if base not in gaze_lookup:
        skipped += 1
        continue

    shutil.copy(
        os.path.join(RGB_DIR, fname),
        os.path.join(IMG_OUT, fname)
    )

    gaze_file = gaze_lookup[base]
    shutil.copy(
        os.path.join(GAZE_DIR, gaze_file),
        os.path.join(COND_OUT, gaze_file)
    )

    captions[fname] = "a realistic scene"

with open(os.path.join(OUT_DIR, "captions.json"), "w") as f:
    json.dump(captions, f, indent=2)

print(f"✔ Gaze training dataset ready | skipped {skipped} images without gaze")
```

```python
import os
import json

BASE_DIR = "/content/drive/MyDrive/controlnet_project/train_gaze"

IMG_DIR = os.path.join(BASE_DIR, "images")
COND_DIR = os.path.join(BASE_DIR, "conditioning")
OUT_FILE = os.path.join(BASE_DIR, "metadata.jsonl")

# Example label mapping (edit if needed)
LABEL_MAP = {
    "0": "looking left",
    "1": "looking right",
    "2": "looking center"
}

with open(OUT_FILE, "w") as f:
    for img_name in sorted(os.listdir(IMG_DIR)):
        if not img_name.lower().endswith((".jpg", ".png")):
            continue

        stem = os.path.splitext(img_name)[0]
        cond_name = stem + ".png"

        # skip if gaze map missing
        if not os.path.exists(os.path.join(COND_DIR, cond_name)):
            continue

        # If labels are numeric filenames:
        raw_label = stem.split("_")[-1]  # adjust if needed
        caption = LABEL_MAP.get(raw_label, raw_label)

        record = {
            "file_name": f"images/{img_name}",
            "conditioning_image": f"conditioning/{cond_name}",
            "label": caption
        }

        f.write(json.dumps(record) + "\n")

print("metadata.jsonl created")
```
